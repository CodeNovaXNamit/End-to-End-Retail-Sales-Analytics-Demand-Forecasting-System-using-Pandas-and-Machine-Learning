```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("../data/processed/cleaned_walmart_sales.csv")
df['Date']=pd.to_datetime(df['Date'])

df.head()
```

```python
weekly_trend= df.groupby('Date')['weekly_sales'].sum()
plt.figure()
weekly_trend.plot()
plt.title("Total Weekly Sales Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sales")
```

Sales exhibit clear seasonal spikes, especially during November and December

```python
df['month']=df['Date'].dt.month
monthly_sales=df.groupby('month')['weekly_sales'].mean()

plt.figure()
monthly_sales.plot(kind='bar')
plt.xlabel("Month")
plt.ylabel("Average Sales")
```

In this graph we can see that in the month of Nov and Dec we have high avg sales because of holiday

```python
holiday_sales=df.groupby('is_holiday')['weekly_sales'].mean()

plt.figure()
holiday_sales.plot(kind='bar')
plt.title("Average Sales: Non-Holiday vs Holiday")
plt.show()
```

This graph shows that on the holiday we got more sales than non holidays

```python
store_sales=df.groupby('store_id')['weekly_sales'].mean().sort_values(ascending=False)

plt.figure()
store_sales.head(10).plot(kind='bar')
plt.title("Top 10 Store by Total Sales")
plt.show()
```

This graph shows that Store 20 is performing the best and the top 9

```python
dept_sales = df.groupby('dept_id')['weekly_sales'].sum().sort_values(ascending=False)

plt.Figure
dept_sales.head(10).plot(kind='bar')
plt.title("Top 10 Departments by Sales")
plt.show()
```

This graph shows that which department is selling the most and its dept 92 and the 9 others
