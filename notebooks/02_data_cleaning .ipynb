```python
import pandas as pd
df=pd.read_csv("../data/raw/walmart_sales.csv")
df.head()
```

```python
df['Date']=pd.to_datetime(df['Date'])
df.info()
```

```python
df.duplicated().sum()
```

This shows that zero duplicated values by using duplicated function and sum.

```python
df=df.sort_values(by='Date')
df.head()
```

```python
df=df.rename(columns={
    'Weekly_Sales':'weekly_sales',
    'IsHoliday':'is_holiday',
    'Store':'store_id',
    'Dept':'dept_id'
})
df.columns
```

I used the renaming feature of pandas to rename the whole columns.

```python
#Basic Sanity checks
df.isna().sum()
df.describe()
```

```python
df.to_csv("../data/processed/cleaned_walmart_sales.csv", index=False)
```
