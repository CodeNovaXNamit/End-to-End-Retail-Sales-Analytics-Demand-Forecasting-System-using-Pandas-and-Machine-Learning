```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/cleaned_walmart_sales.csv")
df['Date'] = pd.to_datetime(df['Date'])

df = df.sort_values(['store_id', 'dept_id', 'Date'])
df.head()
```

Here the sorting is done in this way

first sort on the basis of store_id->dept_id->date

```python
#Time-based feature
df['year']=df['Date'].dt.year
df['month']=df['Date'].dt.month
df['week']=df['Date'].dt.isocalendar().week
df['day_of_week']=df['Date'].dt.dayofweek
df['quarter']=df['Date'].dt.quarter
df.head()
```

By using this our model will undestand following:
1. Monthly Analytics
2. Weekly cycles
3. Quaterly trends

```python
#Lag Feature-what was sales last week
df['lag_1']=df.groupby(['store_id','dept_id'])['weekly_sales'].shift(1)
df['lag_2']=df.groupby(['store_id','dept_id'])['weekly_sales'].shift(2)
df['lag_4']=df.groupby(['store_id','dept_id'])['weekly_sales'].shift(4)
df.head()
```

This will show how much sales in 1 week, 2 week and 1 month

```python
#4.4 Rolling window feature

df['rolling_mean_4']=(
    df.groupby(['store_id','dept_id'])['weekly_sales']
    .shift(1)
    .rolling(window=4)
    .mean()
)

df['rolling_std_4']=(
    df.groupby(['store_id', 'dept_id'])['weekly_sales']
    .shift(1)
    .rolling(window=4)
    .std()
)
df.head()
```
